## 1. 先把要用的东西定义好

这几格只是定义（环境、函数），运行后没有输出也正常。用到的前几章的函数也一起抄进来了，不需要本地文件。

**函数 `episode_return`**（来自 `ch02_return.py`）：rewards[0] 是 R_1。G_t 从 R_{t+1} 加到最后。

```python
def episode_return(rewards, t=0):
    """rewards[0] 是 R_1。G_t 从 R_{t+1} 加到最后。"""
    return sum(rewards[t:])
```

**函数 `all_returns`**（来自 `ch02_return.py`）：下面要用到的小工具。

```python
def all_returns(rewards):
    return [episode_return(rewards, t) for t in range(len(rewards) + 1)]
```

## 2. 一步一步跑

每格对应讲义里的一步，打印的文字说明了在算什么。

```python
dog = [1, 0, 1, 0]
```

```python
print("小狗 G0..G4 =", all_returns(dog))
```

```python
fast, slow = [-1] * 5, [-1] * 8
```

```python
print("迷宫 5 步出去 G0 =", episode_return(fast), "；8 步出去 G0 =", episode_return(slow))
```

## 3. 代码里才看得到的

Python 列表下标从 0 开始，所以 `rewards[0]` 存的是 $R_1$。`episode_return(rewards, t)` 写成 `sum(rewards[t:])`：$G_t$ 从 $R_{t+1}$ 开始，正好对应 `rewards[t]`。讲义的下标和代码的下标差 1，这是 RL 代码里最常见的错位。

## 4. 自己改一改

讲义里建议的几个小实验（数字是实跑出来的，改完对一对）：

1. `all_returns([0, 0, 0, 10])` → `[10, 10, 10, 10, 0]`：奖励只在最后一步给，前面每一步的回报也都是 10。
2. `all_returns([1] * 4)` → `[4, 3, 2, 1, 0]`：每步都有奖励时，越往后剩下的越少。
